```python
import pandas as pd

df=pd.read_csv("Book1.csv")
df.shape # shape of data
df.isnull().sum() # show empty field
df.dropna(how="all") # remove empty
print(df["R_Score"].mean())
df[df["R_Score"] > 10]
df.dropna()
df.dropna(subset=["No_Books"])
df.duplicated().sum() # show duplicated
df[df.duplicated(keep=False)] # show duplicated and original
df = df.drop_duplicates()
df.dtypes
df.describe()  # statistical summary
print("Minimum:", df["R_Score"].min())
print("Range:", df["R_Score"].max() - df["R_Score"].min())
# all numeric median,mean,mode
print("\nMedian:")
print(df.median(numeric_only=True))
```
