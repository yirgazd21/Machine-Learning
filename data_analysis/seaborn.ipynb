```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Book1.csv")

plot_df = df.dropna(subset=["F_Name", "R_Score"])

sns.barplot(
    data=plot_df,
    x="F_Name",
    y="R_Score"
)

plt.title("Researcher Scores")
plt.xlabel("First Name")
plt.ylabel("Research Score")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
