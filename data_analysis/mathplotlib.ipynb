```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("Book1.csv")

# x = [1, 2, 3, 4, 5]
# y = [10, 20, 15, 30, 25]
# plt.plot(x, y)

# plt.plot(df["R_ID"], df["R_Score"], marker="o")
# plt.title("Research Scores")
# plt.xlabel("Researcher Id")
# plt.ylabel("Researcher Scores")
# plt.show()
print(df["R_Score"].describe())
plot_df = df.dropna(subset=["F_Name", "R_Score"])
plt.bar(plot_df["F_Name"], plot_df["R_Score"])

plt.title("Research Scores by Researcher")
plt.xlabel("Researcher Name")
plt.ylabel("Research Score")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# A histogram shows the distribution of numerical data. It groups values into ranges called bins and shows how many values fall within each range.
scores = df["R_Score"].dropna()
plt.hist(
    scores,
    bins=5,
    color="skyblue",
    edgecolor="black"
)

plt.title("Distribution of Research Scores")
plt.xlabel("Research Score")
plt.ylabel("Frequency")

plt.show()
```

```python
plot_df = df.dropna(subset=["No_papers", "R_Score"])
plt.scatter(
    plot_df["No_papers"],
    plot_df["R_Score"],
    color="green",
    s=80,
    marker="o"
)

plt.title("Research Papers vs Research Score")
plt.xlabel("Number of Papers")
plt.ylabel("Research Score")

plt.show()
```

```python
# box plot
scores = df["R_Score"].dropna()

plt.boxplot(scores)

plt.title("Distribution of Research Scores")
plt.ylabel("Research Score")

plt.show()
```
