Creating Df by Function(advanced method)

```python
import pandas as pd
def make_df(cols,ind):
    data={c: [str(c) +str(i) for i in ind] for c in cols}
    return pd.DataFrame(data,ind)
print(make_df('ABC', range(3)))
```

slicing or accessing colmn

```python
w = pd.Series({'a':1 ,'b':2 ,'c':3 ,'d':4 ,'e':5})
x = pd.Series({'a':6 ,'b':7 ,'c':8 ,'d':9 ,'e':10})
y = pd.Series({'a':11 ,'b':12 ,'c':13 ,'d':14 ,'e':15})
z = pd.Series({'a':16 ,'b':17 ,'c':18 ,'d':19 ,'e':20})
grades = pd.DataFrame({'Math':w,'Physics':x,'French':y,'Chemistry':z})
print(grades)
print(grades['Chemistry'])

# To transpose the Dataframe we can use df.T
print(grades.T)

## access on key or value
grades.keys()
grades.values # No bracket need for values

# Applying conditions to specific keys or values
print('Math' in grades.keys()) # cehck existensse
print(12 in grades.values)    # cehck existensse

# Vertical representation of elements with keys and values
print(grades.stack())
```

**Locating specific elements for slicing and searchin**
We have two primary methods for this task:
**Method 1: iloc (i for index)** - It locates the position by index, similar to the method we are familiar with in lists.

Example: --> iloc[:3, :2]

**Method 2: loc** - You need to specify the names of the elements, rows, and columns you are searching for.

Example: --> loc["b":"c", "Math":] This implies selecting rows from 'b' to 'c' and
columns from 'Math' to the end

```python
w = pd.Series({'a':1 ,'b':2 ,'c':3 ,'d':4 ,'e':5})
x = pd.Series({'a':6 ,'b':7 ,'c':8 ,'d':9 ,'e':10})
y = pd.Series({'a':11 ,'b':12 ,'c':13 ,'d':14 ,'e':15})
z = pd.Series({'a':16 ,'b':17 ,'c':18 ,'d':19 ,'e':20})
grades = pd.DataFrame({'Math':w,'Physics':x,'French':y,'Chemistry':z})
print(grades)
print(grades.iloc[:3,:2])
print(grades.iloc[:3,:2])
print(grades.loc['b':'c', 'Math':])
print(grades.loc['b':'d', 'Math':'French'])
print(grades.loc[grades.Math>3,['French','Math']])

#Access names of all columns and the index
print(grades.columns) ## same as df.keys()
print(grades.index)
print(grades['Math'])
```
