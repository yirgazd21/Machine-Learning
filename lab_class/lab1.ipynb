```python
import numpy as np
from sklearn import datasets ,linear_model
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import matplotlib.pyplot as plt


d=datasets.load_diabetes()
x=d.data
y=d.target

print("Feature shape: ", x.shape)
print("Target shape: ", y.shape)

# select BMI only(feature index 2)
x=x[:,2]
x=x.reshape(-1,1)
x.shape  # (442,1)

x_train=x[:-80]
y_train=y[:-80]
x_test=x[-80:]
y_test=y[-80:]

model=linear_model.LinearRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print(y_pred)

r2 =r2_score(y_test,y_pred)
print(f"{r2:.3f}")

# cofficient and intercept

print(F"cofficient: { model.coef_[0]:.3f}")
print(f"Intercept: {model.intercept_:.3f}")

```

```python
plt.scatter(x_test,y_test,color="black", label="data")
plt.plot(x_test,y_pred,color="blue", linewidth=2, label='Regression Line')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Regression')
plt.legend()
plt.show()
```
