```python
import pandas as pd
pd.__version__
data = pd.Series([0.25, 0.5, 0.75, 1.0])
print(data)
data = pd.Series((0.25, 0.5, 0.75, 1.0))
# print(data.values)
# print(data.index)
# print(data.keys)
data.describe()

print(data.agg(['max','min','sum','mean','std']))
## acess and slice data[start:end:step]
print(data[1])
print(data[1:3])
print(data[1:3:2])

## customize index
data1 = pd.Series([1,2,3,4], index=['a', 'b', 'c', 'd'])
data2 = pd.Series({'a':1,'b':2,'c':3,'d':4})
print(data1)
print(data2)

```

## List and list operation

```python
## we can create list using Index
x = pd.Index([2,3,5,7,11])
print(x)
a = pd.Index([1, 3, 5, 7, 9])
b = pd.Index([2, 3, 5, 7, 11])
print(a)
print(b)
print(a & b)
print(a | b)
print(a ^ b)
```
