Advanced operations on DataFrames

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({'A': np.random.rand(10),'B': np.random.rand(10)})
print(df)
print('------------------------------')
print(df.sum())
print('------------------------------')
print(df.prod())
print('------------------------------')
print(df.mean())
print('------------------------------')
print(df['A'].sum())
print('------------------------------')
print(df['B'].prod())
print('------------------------------')
print(df['A'].mean())

# Statics of all columns on DF
print('------------------------------')
print(df.mean(axis='rows'))
print('------------------------------')
print(df.count())
print('------------------------------')
print(df.min())
print('------------------------------')
print(df.max())
print('------------------------------')
print(df.std())
print('------------------------------')
```

Statistics on the entire DF using the Describe method

```python
df = pd.DataFrame({'key':['A','B','C','A','B','C'],
 'data': range(6)},columns=['key', 'data'])
print(df)
print(df.describe())
```

#### Groupby
Groupby is a crucial method in Pandas that facilitates grouping similar
elements and applying various operations on each group.

```python
df = pd.DataFrame({'key':['A','B','C','A','B','C'],
 'data': range(6)},columns=['key', 'data'])
print(df)
print(df.groupby('key').sum())
print(df.groupby('key').describe())

## unstack returned data fro describe
print(df.groupby('key').describe().unstack())

df = pd.DataFrame({'key': ['A', 'B', 'C', 'A', 'B', 'C'],
 'data1': range(6),
 'data2': np.random.randint(0, 10, 6)},
 columns = ['key', 'data1', 'data2'])
print(df)
df2 = df.groupby('key').aggregate({'data1': 'min','data2': 'max'})
print(df2)

# Filter returned data based on function
f = pd.DataFrame({'key': ['A', 'B', 'C', 'A', 'B', 'C'],
 'data1': range(6),
 'data2': np.random.randint(0, 10, 6)},
 columns = ['key', 'data1', 'data2'])
print(df)
def filter_func(x):
 return x['data2'].std() > 4
df2 = df.groupby('key').filter(filter_func)
print(df2)

f = pd.DataFrame({'key': ['A', 'B', 'C', 'A', 'B', 'C'],
 'data1': range(6),
 'data2': np.random.randint(0, 10, 6)},
 columns = ['key', 'data1', 'data2'])
print(df)
df2 = df.groupby('key').transform(lambda x: x**2)
```
