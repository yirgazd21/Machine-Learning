Advanced operations on DataFrames

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({'A': np.random.rand(10),'B': np.random.rand(10)})
print(df)
print('------------------------------')
print(df.sum())
print('------------------------------')
print(df.prod())
print('------------------------------')
print(df.mean())
print('------------------------------')
print(df['A'].sum())
print('------------------------------')
print(df['B'].prod())
print('------------------------------')
print(df['A'].mean())

# Statics of all columns on DF
print('------------------------------')
print(df.mean(axis='rows'))
print('------------------------------')
print(df.count())
print('------------------------------')
print(df.min())
print('------------------------------')
print(df.max())
print('------------------------------')
print(df.std())
print('------------------------------')
```

Statistics on the entire DF using the Describe method

```python
df = pd.DataFrame({'key':['A','B','C','A','B','C'],
 'data': range(6)},columns=['key', 'data'])
print(df)
print(df.describe())
```
