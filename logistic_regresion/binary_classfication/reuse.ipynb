**Load and Reuse for New Predictions (Without Retraining)**

```python
import joblib
import pandas as pd

# 1. Load the saved model
loaded_model = joblib.load('insurance_logistic_model.pkl')

# 2. Make predictions on completely new ages
new_customers = pd.DataFrame({'age': [21, 38, 62]})
predictions = loaded_model.predict(new_customers)
probabilities = loaded_model.predict_proba(new_customers)

for age, pred, prob in zip(new_customers['age'], predictions, probabilities):
    status = "Will Buy" if pred == 1 else "Will NOT Buy"
    confidence = prob[pred] * 100
    print(f"Age {age}: {status} (Confidence: {confidence:.1f}%)")
```
