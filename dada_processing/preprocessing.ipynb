## Step 1(iii): Inspect the dataset

```python

# Step 1(iii): Inspect the dataset

# Display the first 5 rows
print("\nFirst 5 rows:")
print(df.head())

# Display the number of rows and columns
print("\nDataset shape:")
print(df.shape)

# Display column names
print("\nColumn names:")
print(df.columns.tolist())

# Display data types and non-null counts
print("\nDataset information:")
df.info()

# Display statistical summary
print("\nStatistical summary:")
print(df.describe())

# Count missing values in each column
print("\nMissing values:")
print(df.isnull().sum())

# Count duplicate rows
print("\nDuplicate rows:")
print(df.duplicated().sum())
```

## Step 1(iv): Handling Missing Values

```python

# Step 1(iv): Handling Missing Values

# Count missing values in each column
print("\nMissing values before cleaning:")
print(df.isnull().sum())

# Calculate the percentage of missing values
print("\nMissing value percentages:")
print((df.isnull().sum() / len(df)) * 100)




```

```python
# Check dataset size before handling missing values
print("Rows before cleaning:", len(df))

# Count rows with missing glyhb
print("Rows with missing glyhb:", df["glyhb"].isnull().sum())

# Check missing values in blood pressure measurements
print("Missing bp.2s:", df["bp.2s"].isnull().sum())
print("Missing bp.2d:", df["bp.2d"].isnull().sum())


# Remove rows where the target variable glyhb is missing
df = df.dropna(subset=["glyhb"]).copy()

# Check the result
print("\nDataset after removing missing glyhb values:")
print("Rows:", len(df))
print("Missing glyhb values:", df["glyhb"].isnull().sum())
```

```python

# Remove columns with too many missing values
df = df.drop(columns=["bp.2s", "bp.2d"])

# Identify numerical and categorical columns
numeric_cols = df.select_dtypes(include="number").columns
categorical_cols = df.select_dtypes(
    include=["object", "category"]
).columns

# Fill missing numerical values with the median
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Fill missing categorical values with the mode
for col in categorical_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

# Check remaining missing values
print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nDataset shape after cleaning:")
print(df.shape)
```
